```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    "temperature_log.csv",
    header=None,  # because your file doesn’t have a header row
    names=['time', 'temp_hip_1', 'temp_knee_1', 'temp_hip_2', 'temp_knee_2',
           'temp_hip_3', 'temp_knee_3', 'temp_hip_4', 'temp_knee_4', 
           'pos_hip_1', 'pos_knee_1', 'pos_hip_2', 'pos_knee_2', 'pos_hip_3',
           'pos_knee_3', 'pos_hip_4', 'pos_knee_4']
)
print(df.head())
time_0 = df['time'][0]
df['time'] = df['time'] - time_0
df['time_in_hours'] = df['time'] / 3600
df['time_in_minutes'] = df['time'] / 60
hip_color = 'blue'
knee_color = 'red'
plt.figure(figsize=(20, 10))
# plt.plot(df['time_in_minutes'], df['temp_hip_1'], label='hip_1', color=hip_color)
# plt.plot(df['time_in_minutes'], df['temp_hip_2'], label='hip_2', color=hip_color)
# plt.plot(df['time_in_minutes'], df['temp_hip_3'], label='hip_3', color=hip_color)
# plt.plot(df['time_in_minutes'], df['temp_hip_4'], label='hip_4', color=hip_color)
plt.plot(df['time_in_minutes'], df['temp_knee_1'], label='knee 11 (not enclosed)',  marker='o')
plt.plot(df['time_in_minutes'], df['temp_knee_2'], label='knee 21', marker='o')
plt.plot(df['time_in_minutes'], df['temp_knee_3'], label='knee 31 (not enclosed)', marker='o')
plt.plot(df['time_in_minutes'], df['temp_knee_4'], label='knee 41', marker='o')
plt.xlabel('Time (minutes)')
plt.ylabel('Temperature (°C)')
plt.ylim(28, 55)
plt.legend(fontsize=8, loc='lower right')
plt.grid(alpha=0.2)
```
